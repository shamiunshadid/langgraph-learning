```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq

from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()
model = ChatGroq(model="openai/gpt-oss-20b")
```

```python
class BlogState(TypedDict):
    
    title: str
    outline: str
    content: str
```

```python
def create_outline(state: BlogState) -> BlogState:
   
   # fetch title
   title = state['title']
   
   # call llm and generate outline
   prompt = f"Generate an outline for a blog on the topic of {title}"
   outline = model.invoke(prompt).content
   # print("Outline --->", outline)
   # print("-"*10)
   
   # update state
   state['outline'] = outline
   
   return state
   
    
def create_blog(state: BlogState) -> BlogState:
    
    # same fetch title and outline
    title = state['title']
    outline = state['outline']
    
    # call llm and generate the actual blog
    prompt = f"write a blog on the title of {title} using the following outline - {outline}"
    content = model.invoke(prompt).content
    
    # update state
    state['content'] = content
    
    return state
```

```python
graph = StateGraph(BlogState)

# nodes
graph.add_node("create_outline", create_outline)
graph.add_node("create_blog", create_blog)

# edges
graph.add_edge(START, "create_outline")
graph.add_edge("create_outline", "create_blog")
graph.add_edge("create_blog", END)

# compile graph
workflow = graph.compile()
```

```python
# execute graph
initial_state = {"title": "The ongoing ai race"}

final_state = workflow.invoke(initial_state)
print(final_state["outline"])
```
